```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import train_test_split
```

```python
iris = load_iris()
```

```python
x = iris.data
y = iris.target
x_train , x_test, y_train, y_test = train_test_split(x,y, test_size=0.2, random_state=42)
```

```python
model1 = DecisionTreeClassifier(max_depth=1, random_state=42)
model1.fit(x_train, y_train)
pred1 = model1.predict(x_test)
```

```python
model2 = DecisionTreeClassifier(max_depth=3, random_state=42)
model2.fit(x_train, y_train)
pred2 = model2.predict(x_test)
```

```python
model10 = DecisionTreeClassifier(max_depth=10, random_state=42)
model10.fit(x_train, y_train)
pred10 = model10.predict(x_test)
```

```python
print('depth 1', accuracy_score(y_test,pred1))
print('depth 3', accuracy_score(y_test,pred2))
print('depth 10', accuracy_score(y_test,pred10))
```
