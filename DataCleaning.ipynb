```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import StandardScaler, MinMaxScaler
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score, mean_absolute_error
```

```python
df = pandas.DataFrame({
    "name": ["Alice", "Bob", "Charlie", "Diana", "Ethan", "Alice", "Fiona", "George"],
    "age": [25, 30, np.nan, 150, -5, 25, 42, np.nan],
    "salary": [55000, 62000, 72000, np.nan, 48000, 55000, 81000, 67000],
    "gender": ["Female", "Male", "Male", "Female", np.nan, "Female", "Female", "Male"],
    "city": ["New York", "London", "Paris", "Tokyo", "Berlin", "New York", np.nan, "London"],
    "experience": [2, 5, 8, 20, np.nan, 2, 15, 7]
})

# Add a duplicate row
df = pandas.concat([df, df.iloc[[1]]], ignore_index=True)

```

```python
df.head()
df.tail()
df.shape
df.columns
df.dtypes
df.info()
df.describe()
df
```

```python
df.isnull()
df.isnull().sum()
```

```python
df = df.fillna(df.mean(numeric_only=True))
df.isnull()
```

```python
df.duplicated()
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df[(df['age'] < 0) | (df['age'] > 100)]

df.loc[(df['age'] < 0) | (df['age'] > 100), 'age'] = None
df['age'] = df['age'].fillna(df['age'].median())
df[(df['age'] < 0) | (df['age'] > 100)]
```

```python
df['gender'] = df['gender'].str.strip().str.capitalize().map({'Male':0,'Female':1})

df.head()
```

```python
df_standerd = df.copy()
```

```python
df_standerd[['age', ' salary', 'experience']] = StandardScaler().fit_transform(df[['age','salary', 'experience']])
df_standerd.head()
```

```python
df_minmax = df.copy()
df_minmax[['age','salary', 'experience']] = MinMaxScaler().fit_transform(df[['age','salary', 'experience']])
df_minmax.head()
```

```python
df['gender'].isnull().sum()
df['gender'] = df['gender'].fillna(df['gender'].mode()[0])
```

```python
df = pd.get_dummies(df, columns=['city'], dtype=int)
```

```python
x = df[['age' , 'experience']]
y = df['salary']
```

```python
x_train,x_test ,y_train, y_test = train_test_split(x,y,test_size=0.2, random_state=42)
```

```python
model = LinearRegression()
```

```python
model.fit(x_train, y_train)
```

```python
prediction = model.predict(x_test)
```

```python
print(r2_score(y_test,prediction))
print(mean_absolute_error(y_test,prediction))
```
