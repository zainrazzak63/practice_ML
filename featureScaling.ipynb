```python
import pandas as pd
from sklearn.preprocessing import StandardScaler, MinMaxScaler
```

```python

data = {
    "age": [22, 25, 28, 32, 35, 38, 42, 45, 50, 55],
    "salary": [30000, 35000, 42000, 50000, 58000, 65000, 75000, 85000, 95000, 110000],
    "experience": [0, 2, 4, 7, 10, 13, 17, 20, 25, 30]
}

df = pd.DataFrame(data)
```

```python
features =df[['age', 'salary', 'experience']]
```

```python
scaler = StandardScaler()
scaled_data = scaler.fit_transform(features)
```

```python
df_standerd = pd.DataFrame(scaled_data, columns=['age','salary', 'experience'])
print(df_standerd)
```

```python
scaler = MinMaxScaler()
```

```python
minmax_data = scaler.fit_transform(features)
```

```python
df_minmax = pd.DataFrame(
    minmax_data, columns=['age','salary','experience']
)
print(df_minmax)
```
